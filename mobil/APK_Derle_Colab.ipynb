## 4. APK Derle

⚠️ **Bu adım ~20 dakika sürebilir.** İlk seferde Android SDK ve NDK indirilir.

````python
!yes | buildozer android debug 2>&1 | tail -50

import glob
apks = glob.glob("bin/*.apk")
if apks:
    print(f"\n\u2705 APK basariyla derlendi!")
    for apk in apks:
        size = os.path.getsize(apk) / (1024 * 1024)
        print(f"   {apk} ({size:.1f} MB)")
else:
    print("\n\u274c APK olusturulamadi! Yukaridaki loglari kontrol edin.")
````

## 5. APK'yı İndir

````python
from google.colab import files
import glob

apks = glob.glob("bin/*.apk")
if apks:
    for apk in apks:
        print(f"Indiriliyor: {apk}")
        files.download(apk)
else:
    print("APK bulunamadi. Once 4. adimi calistirin.")
````
